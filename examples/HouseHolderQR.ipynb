```python
import numpy as np 
from lineq.qr_decomposition import qr_decomposition
from lineq.gauss import backward_substitution
```

```python
A = np.array([
        [3.,0.,1.,0.],
        [0.,1.,0.,1.],
        [0.,0.,2.,1.],
        [1.,0.,1.,0.],
        
])
b = np.array([[1.],[2.],[2.],[4.]])
```

```python
Q, R = qr_decomposition(A)
b_prim = Q.T @ b
x = backward_substitution(R, b_prim)
x
```
