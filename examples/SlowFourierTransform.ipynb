```python
import numpy as np
import matplotlib.pyplot as plt
from interpolation.slow_fourier import slow_fourier_transform, construct_function
```

```python
samples = np.array([1, 2, 4, 8, 16, 1])
```

```python
c_k = slow_fourier_transform(samples)
funct = construct_function(c_k)
```

```python
ts = np.arange(0, 2 * np.pi, 0.02)
sample_percentage = np.arange(0, len(samples), 1)/len(samples)
plt.scatter(2 * np.pi * sample_percentage, samples)
plt.plot(ts, np.real(funct(ts)))
```
