```python
import numpy as np

from iteration.vector import power_iteration, get_eigenvalue
```

```python
A = np.array([
    [1., 0.],
    [2., 2.]
])
```

```python
eigenvector1 = A @ power_iteration(A, steps=100)
eigenvalue1 = get_eigenvalue(A, eigenvector1)
eigenvector1, eigenvalue1, A @ eigenvector1
```

```python
eigenvector2 = power_iteration(A, gamma = 2., steps=100)
eigenvalue2 = get_eigenvalue(A, eigenvector2)
eigenvector2, eigenvalue2, A @ eigenvector2
```
