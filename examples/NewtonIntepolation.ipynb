```python
import numpy as np
import matplotlib.pyplot as plt
from interpolation.newton import NewtonTable
```

```python
xs = [2., 3.]
ys = [5., 10.]
table = NewtonTable(xs, ys)
print(table.polynom)
```

```python
ts = np.arange(table.xs[0], table.xs[-1], 0.002)
plt.scatter(table.xs, table.cs[:,0])
plt.plot(ts, table(ts))
```

```python
new_xs = [-3, -2,  0.]
new_ys = [10., 5., 1.]
table.add_points_before(new_xs, new_ys)
print(table.polynom)
```

```python
ts = np.arange(table.xs[0], table.xs[-1], 0.002)
plt.scatter(table.xs, table.cs[:,0])
plt.plot(ts, table(ts))
```
