```python
import numpy as np
import matplotlib.pyplot as plt

from interpolation.lagrange import LagrangePolynoms
```

```python
xs = np.array([0., 1.,3., 4.])
ys = np.array([1. ,4.,-2., 2.])
```

```python
polynom = LagrangePolynoms(xs, ys)
polynom
```

```python
polynom.to_standard_polynom()
```

```python
ts = np.arange(xs[0], xs[-1], 0.02)
plt.scatter(xs, ys)
plt.plot(ts, polynom(ts))
```
